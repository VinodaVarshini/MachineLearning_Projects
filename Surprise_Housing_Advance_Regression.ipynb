## 1. Problem Statement

A US-based housing company named Surprise Housing has decided to enter the Australian market. The company uses data analytics to purchase houses at a price below their actual values and flip them on at a higher price. For the same purpose, the company has collected a data set from the sale of houses in Australia.

The company is looking at prospective properties to buy to enter the market. You are required to build a regression model using regularisation in order to predict the actual value of the prospective properties and decide whether to invest in them or not.

The company wants to know:

- Which variables are significant in predicting the price of a house
- How well those variables describe the price of a house.

### Business Goal

You are required to model the price of houses with the available independent variables. This model will then be used by the management to understand how exactly the prices vary with the variables. They can accordingly manipulate the strategy of the firm and concentrate on areas that will yield high returns. Further, the model will be a good way for management to understand the pricing dynamics of a new market.

### Which are the features having maximum single value?

```python
# Frequently occuring single value from categorical columns
def housing_occurance(category_features):
    hou_occur = []
    for i in category_features.columns:
        occurance_count = category_features[i].value_counts()
        count_total = occurance_count.iloc[0]
        if count_total / len(category_features) * 100 > 98:
            hou_occur.append(i)
    hou_occur = list(hou_occur)
    return hou_occur
```

- __Street, Utilities, Condition2, RoofMatl, PoolQC__ has maximum number of records with __single value which is more than 98%__.

```python
## Define a function
def mlr_result(y_train, y_train_pred, y_test, y_pred):

    metric1 = []

## Calculatig R-squared value
    print("R-Squared-Train =", '%.2f' % r2_score(y_train, y_train_pred))
    metric1.append(r2_score(y_train, y_train_pred))
    print("R-Squared-Test =", '%.2f' % r2_score(y_test, y_pred))
    metric1.append(r2_score(y_test, y_pred))

## Calculating Residual sum of squares
    rss_train = np.sum(np.square(y_train - y_train_pred))
    metric1.append(rss_train)
    rss_test = np.sum(np.square(y_test - y_pred))
    metric1.append(rss_test)
    print("RSS-Train =", '%.2f' % rss_train)
    print("RSS-Test =", '%.2f' % rss_test)

## Calculating Mean Squared Error
    mse_train = mean_squared_error(y_train, y_train_pred)
    metric1.append(mse_train)
    mse_test = mean_squared_error(y_test, y_pred)
    metric1.append(mse_test)
    print("MSE-Train =", '%.2f' % mse_train)
    print("MSE-Test =", '%.2f' % mse_test)

# Calulating Root Mean Squared Error
    rmse_train = mse_train**0.5
    metric1.append(rmse_train)
    rmse_test = mse_test**0.5
    metric1.append(rmse_test)
    print("RMSE-Train =", '%.2f' % rmse_train)
    print("RMSE-Test =", '%.2f' % rmse_test)


    return metric1
```
